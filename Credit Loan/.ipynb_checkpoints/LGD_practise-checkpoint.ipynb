```python
df1 = pd.get_dummies(data['Gender'],dtype=int,drop_first=True)
df2 = pd.get_dummies(data['Married'],dtype=int,drop_first=True)

df_new = pd.concat([data,df1,df2],axis=1)
df_new.head()
```

```python
df_new = df_new.drop(columns=['Gender','Married'],axis=1)
```

## Split Data

```python
# lets build our linear model
# independant variables
X = df_new.drop(['Losses in Thousands'], axis=1)
# the dependent variable
y = df_new[['Losses in Thousands']]

# Split X and y into training and test set in 70:30 ratio

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=1)

# Fit Linear Model
regression_model = LinearRegression()
regression_model.fit(X_train, y_train)

#Print regression coefficient
print(regression_model.coef_)

#output the intercept
print(regression_model.intercept_)

#Predict for y
y_pred_reg = regression_model.predict(X_test)

#Xtrain and Xtest score
print(f"regression model Train score: {round(regression_model.score(X_test, y_test),4)}")
print(f"regression model test score: {round(regression_model.score(X_test, y_test),4)}")
```

```python
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred_reg))
print("Mean Absolute error: %.2f" % mean_absolute_error(y_test, y_pred_reg))
# The coefficient of determination: 1 is perfect prediction
print("Coefficient of determination: %.2f" % r2_score(y_test, y_pred_reg))
```

## Adding interaction terms 

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn import linear_model

poly = PolynomialFeatures(degree=2, interaction_only=True)
X_train2 = poly.fit_transform(X_train)
X_test2 = poly.fit_transform(X_test)

poly_clf = linear_model.LinearRegression()

poly_clf.fit(X_train2, y_train)

y_pred = poly_clf.predict(X_test2)

#print(y_pred)

#In sample (training) R^2 will always improve with the number of variables!
print(poly_clf.score(X_train2, y_train))
```

```python
#Xtrain and Xtest score
print(f"Poly model Train score: {round(poly_clf.score(X_train2, y_train),4)}")
print(f"Poly model test score: {round(poly_clf.score(X_test2, y_test),4)}")
```

```python
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred))
print("Mean Absolute error: %.2f" % mean_absolute_error(y_test, y_pred))
# The coefficient of determination: 1 is perfect prediction
print("Coefficient of determination: %.2f" % r2_score(y_test, y_pred))
```

# Create a regularized RIDGE model and note the coefficients

```python
ridge = Ridge(alpha=.3)
ridge.fit(X_train2,y_train)
print ("Ridge model:", (ridge.coef_))
```

```python
# Create a regularized LASSO model and note the coefficients
lasso = Lasso(alpha=0.3)
lasso.fit(X_train2,y_train)
print ("Lasso model:", (lasso.coef_))

# Observe, many of the coefficients have become 0 indicating drop of those dimensions from the model
```
